## Dummy Example

```python
from pydantic import BaseModel, Field
class ResponseFormatter(BaseModel):
    """Always use this tool to structure your response to the user."""
    answer: str = Field(description="The answer to the user's question")
    followup_question: str = Field(description="A followup question the user could ask")
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2")

model_with_tools = llm.bind_tools([ResponseFormatter])

ai_msg = model_with_tools.invoke("What is the powerhouse of the cell?")
```

```python
ai_msg.tool_calls[0]["args"]
```

# Task output structure

```python
from typing import List, Literal
from pydantic import BaseModel, Field

class TaskResponseFormatter(BaseModel):
    """Always use this tool to structure your response to the user in the planning phase."""
    task: Literal["summary_statistics", "predict_population_linear", "calculator", "plot_population"] = Field(description="Name of the tool to run")
    id: str = Field(description="Unique identifier for the task")
    dep: List[str] = Field(description="List of task ids this task depends on")
    args: str = Field(description="Arguments for the tool")
```

```python
from langchain_ollama import ChatOllama

llm2 = ChatOllama(model="llama3.2")

model_with_tools2 = llm2.bind_tools([TaskResponseFormatter])
```

```python
ai_msg2 = model_with_tools2.invoke("Please plot the population over the years from the dataset and then perform a forecast for 2028.")
ai_msg2.tool_calls[0]["args"]
```

## Pandas Dataframe

This notebook shows how to use agents to interact with a Pandas DataFrame...this agent calls the Python agent under the hood, which executes LLM generated Python code - this can be bad if the LLM generated Python code is harmful.

```python
from langchain.agents.agent_types import AgentType
from langchain_experimental.agents import create_pandas_dataframe_agent
from langchain_openai import ChatOpenAI
from langchain_community.llms import Ollama
import pandas as pd
```

```python
import pandas as pd
from langchain_openai import OpenAI

df = pd.read_csv(
    "https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv"
)
```

```python
llm = Ollama(model="llama2")
```

```python
agent = create_pandas_dataframe_agent(
    llm=llm,
    df=df,
    verbose=True,
)
```
